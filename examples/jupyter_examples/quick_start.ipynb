## 2. Basic Usage - Generate Captions

```python
from auto_caption import CaptionGenerator

# Initialize generator
generator = CaptionGenerator(model_name="base")

# Generate captions from your video
video_path = "your_video.mp4"  # Change this to your video path
result = generator.generate_captions(video_path)

# Print the transcription
for segment in result['segments']:
    print(f"[{segment['start']:.1f}s - {segment['end']:.1f}s] {segment['text']}")
```

## 3. Add Emotion Detection

```python
from auto_caption import EmotionDetector

# Detect emotions from video
detector = EmotionDetector()
emotions = detector.process_video(video_path)

# Show emotion timeline
print("Detected emotions:")
for emotion in emotions[:5]:  # Show first 5
    print(f"Time: {emotion['timestamp']:.1f}s - {emotion['emotion']} ({emotion['confidence']:.2f})")
```

## 4. Style Captions Based on Emotions

```python
from auto_caption import CaptionStyler, EmotionCategory

# Create a styler
styler = CaptionStyler()

# Style some text based on emotion
text = "This is amazing!"
styled = styler.style_caption(
    text=text,
    emotion=EmotionCategory.EXCITED,
    confidence=0.9
)

print(f"Original: {text}")
print(f"Styled: {styled['formatted_text']}")
```

## 5. Generate Styled Subtitle File

```python
from auto_caption.subtitle import ASSGenerator
import json

# Create caption data with emotions
caption_data = {
    "video_file": video_path,
    "duration": 10.0,
    "segments": [
        {
            "start": 0.0,
            "end": 3.0,
            "text": "Hello everyone!",
            "emotion": "happy",
            "formatted_text": "Hello everyone!"
        },
        {
            "start": 3.0,
            "end": 6.0,
            "text": "Check this out!",
            "emotion": "excited",
            "formatted_text": "CHECK THIS OUT!!!"
        }
    ]
}

# Generate ASS subtitle file
ass_gen = ASSGenerator()
subtitle_file = ass_gen.generate_ass_file(
    caption_data=caption_data,
    output_path="styled_subtitles.ass"
)

print(f"Subtitle file created: {subtitle_file}")
```

## 7. Visualize Results

```python
import matplotlib.pyplot as plt
from collections import Counter

# Count emotions if we have them
if 'emotions' in locals() and emotions:
    emotion_counts = Counter(e['emotion'] for e in emotions)
    
    # Create a bar chart
    plt.figure(figsize=(10, 6))
    plt.bar(emotion_counts.keys(), emotion_counts.values())
    plt.xlabel('Emotions')
    plt.ylabel('Count')
    plt.title('Emotion Distribution in Video')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```
